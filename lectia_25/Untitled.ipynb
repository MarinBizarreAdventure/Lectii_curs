```python
data = data[data['gender'].isin(['Female','Male'])]
data['gender'].value_counts()
```

```python
bmi_mean = data['bmi'].mean()
data['bmi'] = data['bmi'].apply(lambda x: bmi_mean if pd.isnull(x) else x)

data.isnull().sum()
```

```python
histogram_columns = ['age','bmi', 'avg_glucose_level']
data[histogram_columns].hist(figsize=(15,8), bins= 30)
```

```python
data.info()
```

```python
numerical_columns = ['age', 'avg_glucose_level','bmi']

fig, (ax1,ax2,ax3) = plt.subplots(nrows=3, ncols=1, figsize=(10,15))

sns.histplot(x="age", hue="stroke", data=data, kde=True, ax=ax1, bins=30)
ax1.set_title("Histogram age vs Stroke")
ax1.set_xlabel("age")
ax1.set_ylabel("Count")

sns.histplot(x="avg_glucose_level", hue="stroke", data=data, kde=True, ax=ax2, bins=30)
ax2.set_title("Histogram avg_glucose_level vs Stroke")
ax2.set_xlabel("avg_glucose_level")
ax2.set_ylabel("Count")

sns.histplot(x="bmi", hue="stroke", data=data, kde=True, ax=ax3, bins=30)
ax3.set_title("Histogram bmi vs Stroke")
ax3.set_xlabel("bmi")
ax3.set_ylabel("Count")

plt.tight_layout()
plt.show()
```

```python
stroke_heart_disease_count = pd.crosstab(data['heart_disease'], data['stroke'])

stroke_heart_disease_count.div(stroke_heart_disease_count.sum(axis=1),axis=0).plot(kind="bar",stacked=True)
plt.title("Proportion of people with heart disease who had a stroke")
plt.ylabel("proportion")
plt.xlabel('heart Disease')
plt.xticks(ticks=[0,1], labels=['no heart disease', 'heart disease'])
plt.legend(title='stroke')
plt.show()
```
